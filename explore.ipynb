```python
import pandas as pd
import json
from glob import glob

pd.set_option('display.max_columns', None)
```

```python
file = 'data_msk/msk_2026-5.json'
with open(file) as f:
    json_str = json.load(f)
```

```python
json_str = []
for file in glob("data_msk/*.json"):
    with open(file) as f:
        json_str.append(json.load(f))
```

```python
print(len(json_str))
```

# dtp

```python
record_path = ['results', 'region_list', 'pok_list', 'result', 'dtpcardlist', 'info_dtp']
meta = [
    ['results', 'region_list', 'reg_code'],
    ['results', 'region_list', 'reg_name'],
    ['results', 'region_list', 'pok_list', 'pok_code'],
    ['results', 'region_list', 'pok_list', 'pok_name'],
    # ['results', 'region_list', 'pok_list', 'result', 'countcard'],
    # ['results', 'region_list', 'pok_list', 'result', 'pog_v'],
    # ['results', 'region_list', 'pok_list', 'result', 'ran_v']
]
df_dtp = pd.json_normalize(
    json_str,
    record_path=record_path,
    meta=meta,
    errors='ignore'
)
df_dtp = df_dtp.rename(columns={
    'results.region_list.reg_code': 'reg_code',
    'results.region_list.reg_name': 'reg_name',
    'results.region_list.pok_list.pok_code': 'pok_code',
    'results.region_list.pok_list.pok_name': 'pok_name',
})
```

```python
print(df_dtp.columns)
```

```python
df_dtp.head()
```

# ts

```python
# Нормализуем ts_info отдельно, добавив номер ДТП как ключ
df_ts = pd.json_normalize(
    json_str,
    record_path=['results', 'region_list', 'pok_list', 'result', 'dtpcardlist', 'info_dtp', 'ts_info'],
    meta=[['results', 'region_list', 'pok_list', 'result', 'dtpcardlist', 'info_dtp', 'empt_number']]
)
df_ts = df_ts.rename(columns={
    'results.region_list.pok_list.result.dtpcardlist.info_dtp.empt_number': 'empt_number',
})
```

```python
print(df_ts.columns)
```

```python
df_ts[6:20]
```

# uch

```python
df_uch = pd.json_normalize(
    json_str,
    record_path=['results', 'region_list', 'pok_list', 'result', 'dtpcardlist', 'info_dtp', 'ts_info', 'ts_uch'],
    meta=[['results', 'region_list', 'pok_list', 'result', 'dtpcardlist', 'info_dtp', 'empt_number'],
          ['results', 'region_list', 'pok_list', 'result', 'dtpcardlist', 'info_dtp', 'ts_info', 't_ts']
         ],
    errors='ignore'
)
df_uch = df_uch.rename(columns={
    'results.region_list.pok_list.result.dtpcardlist.info_dtp.empt_number': 'empt_number',
    'results.region_list.pok_list.result.dtpcardlist.info_dtp.ts_info.t_ts': 't_ts',
})
```

```python
df_uch.columns
```

```python
df_uch[6:20]
```
